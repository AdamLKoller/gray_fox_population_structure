```python
import pandas as pd
```

```python
dfNonJittered = pd.read_csv('../config/meta_apriori.csv', index_col=0)
dfJittered = pd.read_csv('../config/meta_apriori_jittered.csv', index_col=0)
# Rename loatitude and longitude columns in dfJittered
dfJittered.rename(columns={'latitude': 'latitude_jittered', 'longitude': 'longitude_jittered'}, inplace=True)
# Merge the two DataFrames on the index (assuming they have the same index)
dfMerged = pd.merge(dfNonJittered, dfJittered[['latitude_jittered', 'longitude_jittered']], left_index=True, right_index=True)
dfMerged
```

```python
raw_df = pd.read_csv('../tables/raw_reads.csv')
deduped_df = pd.read_csv('../tables/deduped_reads.csv')
trimmed_df = pd.read_csv('../tables/trimmed_reads.csv')
aligned_df = pd.read_csv('../tables/aligned_reads.csv').drop(['aligned_reads'], axis=1)
reads_df = raw_df.merge(deduped_df, on='Sample_ID').merge(trimmed_df, on = 'Sample_ID').merge(aligned_df, on='Sample_ID')
reads_df
```

```python
dfReadsAndMetadata = dfMerged.merge(reads_df, on='Sample_ID')
dfReadsAndMetadata
```
