### Importing modules

```python
import pandas as pd
```

### Loading STRUCTURE q matrices

```python
q_2 = pd.read_csv('../data/structure/q_matrix_2', index_col=0)
q_2 = q_2.apply(pd.to_numeric)
q_2['max_prob'] = q_2.iloc[:, 1:].max(axis=1)
q_2['cluster_assignment'] = q_2.iloc[:, 1:-1].idxmax(axis=1)
q_2['Cluster'] = q_2['cluster_assignment'].apply(lambda x: x.split('_')[-1])
q_2
```

```python
q_2.groupby('Cluster').count()
```

```python
q_3 = pd.read_csv('../data/structure/q_matrix_3', index_col=0)
q_3 = q_3.apply(pd.to_numeric)
q_3['max_prob'] = q_3.iloc[:, 1:].max(axis=1)
q_3['cluster_assignment'] = q_3.iloc[:, 1:-1].idxmax(axis=1)
q_3['Cluster'] = q_3['cluster_assignment'].apply(lambda x: x.split('_')[-1])
q_3
```

```python
q_3.groupby('Cluster').count()
```
